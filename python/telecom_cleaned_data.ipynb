```python
# Change the datatype of signup_date (Object => datetime)

df["signup_date"] = pd.to_datetime(df["signup_date"], errors = "coerce").fillna("Invalid Date")

df.info()
```

```python
# I have created this column for which customer who has raise a support_ticket or not
df['has_support_ticket'] = df['avg_satisfaction_score'].notnull().map({True: 'Yes', False: 'No'})
# avg_satisfaction_score ko fillna mat karo - blank/NaN hi rehne do

# usage columns pe 0 hi theek hai (genuinely zero, numeric issue nahi hoga)
df['avg_satisfaction_score'] = df['avg_satisfaction_score'].fillna(0)
df['avg_monthly_data_gb'] = df['avg_monthly_data_gb'].fillna(0)
df['avg_monthly_call_minutes'] = df['avg_monthly_call_minutes'].fillna(0)
```

```python
df.head(5)
```

```python
# 1. Tenure ko readable buckets mein toda (dashboard filter ke liye)
def tenure_bucket(months):
    if months <= 12:
        return "0-1 Year"
    elif months <= 24:
        return "1-2 Years"
    elif months <= 36:
        return "2-3 Years"
    elif months <= 48:
        return "3-4 Years"
    else:
        return "4+ Years"

df['tenure_bucket'] = df['tenure_months'].apply(tenure_bucket)
```

```python
# 2. Revenue at risk - agar customer churn kar jaye to kitna annual revenue loss hoga
df['annual_revenue_at_risk'] = np.where(df['churn'] == 'Yes', df['monthly_charges'] * 12, 0)
```

```python
# 4. Total services subscribed count (kitne add-ons liye hain - engagement signal)
addon_cols = ['online_security','online_backup','device_protection',
              'tech_support','streaming_tv','streaming_movies']
df['total_addons_subscribed'] = df[addon_cols].apply(
    lambda row: sum(1 for v in row if v == 'Yes'), axis=1
)
```

```python
# 5. Churn ko readable binary flag bhi bana lo (dashboard KPI cards ke liye easy)
df['is_churned'] = df['churn'].map({'Yes': 1, 'No': 0})
```
